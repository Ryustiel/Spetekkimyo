```python
from pathlib import Path
import re
from html import unescape
import uharfbuzz as hb
from fontTools.ttLib import TTFont

font_path = Path(notebook_dir) / 'output' / 'test.otf'
tt_font = TTFont(str(font_path))

# Enable every feature tag compiled into the font.
feature_tags = []
for table_tag in ('GSUB', 'GPOS'):
    if table_tag in tt_font:
        table = tt_font[table_tag].table
        if table.FeatureList:
            feature_tags.extend(record.FeatureTag for record in table.FeatureList.FeatureRecord)
feature_tags = list(dict.fromkeys(feature_tags))
requested_features = {tag: True for tag in feature_tags}
feature_css = ', '.join(f"'{tag}' 1" for tag in feature_tags) or 'normal'

html = """<style>p {font-size: 120px;} @font-face {src:url('./output/test.otf?version=%s');font-family:'test';} .ss {font-family:'test'; font-feature-settings: %s;}</style>
<p>%s</p>""" % (counter, feature_css, e)
display(HTML(html))

hb_face = hb.Face(font_path.read_bytes())
hb_font = hb.Font(hb_face)
hb.ot_font_set_funcs(hb_font)

# Shape each custom-font run separately to match browser run boundaries.
font_runs = [
    unescape(match.group(1))
    for match in re.finditer(r'<span class="ss">(.*?)</span>', e, flags=re.S)
]
if not font_runs:
    font_runs = [unescape(re.sub(r'<[^>]+>', '', e))]

for run_index, run_text in enumerate(font_runs, 1):
    messages = []

    def _on_message(message):
        messages.append(message)
        return True

    buffer = hb.Buffer()
    buffer.add_str(run_text)
    buffer.guess_segment_properties()
    buffer.set_message_func(_on_message)
    hb.shape(hb_font, buffer, requested_features)

    glyph_names = [tt_font.getGlyphName(info.codepoint) for info in buffer.glyph_infos]

    print(f'Run {run_index}:', run_text)
    print('Final glyphs:', glyph_names)
    print('Lookup trace:')
    for message in messages:
        print(message)
    print()
```
